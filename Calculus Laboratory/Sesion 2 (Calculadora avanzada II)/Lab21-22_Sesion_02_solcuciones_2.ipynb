### **Ejercicio 3** (EVAU 2021, Matemática Aplicadas a las Ciencias Sociales II) ###

Se considera la matriz 

$$
A = \begin{pmatrix}
a&0&1\\
0&b&0\\
1&0&a\\
\end{pmatrix}.
$$

**a)** Determine los valores de los parámetros reales $a$ y $b$ para los que $A = A^{-1}$.

**b)** Para $a = b = 2$, calcule la matriz inversa de $A$.

*Sugerencia:*  En SageMath, la instrucción $\texttt{A[i,j]}$ devuelve el elemento que ocupa la posición $(i,j)$ en la matriz $A$. Es muy importante tener en cuenta que SageMath empieza a numerar las filas y columnas por el índice 0.

-----------
> En lugar de resolver para la igualdad $A=A^{-1}$, tomamos la igualdad equivalente $A^2=\verb|matriz identidad de tamaño 3|$. Esta última se genera con $\verb|identity|\_\verb|matrix(3)|$.

> Necesitamos _informar_ al intérprete de las variables simbólicas $\verb|a|$ y $\verb|b|$ y trabajar con una matriz con entradas en el _anillo simbólico_ ($\texttt{SR}$).

```python
var('a b')
A=matrix(SR,3,[a,0,1,0,b,0,1,0,a])
AA=A^2
show(A,AA)
```

> Con los datos ya preparados, nos disponemos a resolver:

```python
#a)
solve([(AA-identity_matrix(3))[i,j] for i in range(3)  for j in range(3)],a,b)
```

> Con el método $\verb|.subs|()$ (o su _alias_ $\verb|.substitute|()$) podemos dar valores concretos a los parámetros:

```python
#b)
B=(A.subs(a=2).subs(b=2))
invB=B^-1
show(B,invB)
```

> Obsérvese que aunque hemos utilizado el método $\verb|.subs()|$ sobre la matriz $\verb|A|$, esta no se ha modificado.

```python
show(A)
```

### **Ejercicio 4** (EVAU 2021, Matemáticas II) ###


Una tienda online de productos gourmet elabora tres tipos de cafés exclusivos, el Gold Cuvée (a 7.85 euros/kg),
el Paradiso (a 13.3 euros/kg) y el Cremissimo (a 24.85 euros/kg). Para ello utiliza solo dos tipos de grano, el
Arábica y el Robusta. El Gold Cuvée tiene un 90% de grano tipo Arábica, el Paradiso un 85% y el Cremissimo
un 80%.
A lo largo de un mes han necesitado utilizar 27.1 kg de grano del tipo Robusta para atender todos los pedidos y
han ingresado un total de 3112.5 euros. Sabiendo que se ha vendido doble cantidad de café Cremissimo que de
las otras dos especialidades juntas, se pide calcular los kilogramos de grano del tipo Arábica que se han utilizado
a lo largo de ese mes.

-------
> En el cuadro que sigue hemos optado por escribir y guardar en variables las tres ecuaciones a resolver, y pasar sus nombres en una lista al comando $\verb|solve|$. Así el código queda más legible.


```python
var('GC P C')
ec1=(0.1*GC+0.15*P+0.2*C==27.1)
ec2=(7.85*GC+13.3*P+24.85*C==3112.5)
ec3=(C==2*GC+2*P)
solve([ec1,ec2,ec3],GC,P,C)
```

> Si queremos utilizar los valores que resuelven el sistema, ya vimos en clase el uso de la opción $\verb|solution|\_\verb|dict=True|$. Aprovechamos este ejercicio para ilustrar de nuevo este uso.

```python
solucion=solve([ec1,ec2,ec3],GC,P,C,solution_dict=True)
print(solucion)
gc,p,c=(solucion[0][v] for v in [GC,P,C])
gc+p+c-27.1
```

>Una breve explicación de la línea $\verb|gc,p,c=...|$: 

>>Aquí estamos haciendo una triple asignación. En la izquierda aparecen tres nombres, y a la derecha una _tupla_ con tres valores: $\verb|solucion[0][GC]|$ que se asigna a $\verb|gc|$; $\verb|solucion[0][P]|$, asignado a $\verb|p|$; y $\verb|solucion[0][C]|$, para $\verb|c|$. Decimos de este tipo de asignaciones que estamos _desempaquetando_ la tupla.

> Como alternativa, podemos echar mano de la potencia del cálculo matricial: resolvemos $A\mathbf{x}=B$ con $A$ la matriz de coeficientes del sistema lineal, $\mathbf{x}$ el vector de incógnitas, y $B$ la matriz, de una columna, de términos independientes. La solución me la daría tanto $\verb|A\B|$ como $\verb|A^-1*B|$.

```python
A=matrix(3,[.1,.15,.2,7.85,13.3,24.85,2,2,-1])
B=matrix(3,[27.1,3112.5,0])
A\B, A^-1*B
```

> Como vemos, en este caso, al darle las entradas con decimales, la operación "\\" para resolver arrastra ciertos, aunque mínimos, errores de cálculo. 

> Podemos forzar a que trabaje con aritmética exacta, escribiendo los elementos de las matrices como fracciones. Así, el intérprete realizará las cuentas en los racionales.

```python
A=matrix(3,[1/10,15/100,2/10,785/100,133/10,2485/100,2,2,-1])
B=matrix(3,[271/10,31125/10,0])
A\B,A^-1*B
```

> Otra opción es dejar la notación con decimales, pero forzando a que el intérprete los tome como números en $\texttt{QQ}$. Esto, que es optativo, se hace en la primera posición del constructor $\verb|matrix|$.

```python
A=matrix(QQ,3,[.1,.15,.2,7.85,13.3,24.85,2,2,-1])
B=matrix(QQ,3,[27.1,3112.5,0])
A\B, A^-1*B
```
