```python
import numpy as np
import random
import matplotlib.pyplot as plt

data = []
for i in range(10000):
    x = random.uniform(-10, 10)
    y = random.uniform(-10, 10)
    val = None
    if x**2 + y**2 < 40:
        val = 1
    else:
        val = 0
    data.append([(x,y), val])
data = np.array(data, dtype=object)
data_inside = [x for (x,v) in data if v == 1]
data_outside = [x for (x,v) in data if v != 1]

plt.scatter([x for (x,y) in data_inside], [y for (x,y) in data_inside], color = "red", s=1)
plt.scatter([x for (x,y) in data_outside], [y for (x,y) in data_outside], color = "blue", s=1)
plt.gca().set_aspect('equal')

plt.show()
```
