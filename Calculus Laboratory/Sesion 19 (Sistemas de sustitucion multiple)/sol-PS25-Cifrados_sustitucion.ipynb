**(a)** Implementar una función de Sage, _Encripta1()_, que espere una cadena de caracteres y un par de enteros positivos y devuelva la cadena encriptada siguiendo el algoritmo anterior, con el alfabeto del ejemplo. Si los enteros no permiten avanzar en el algoritmo, la función devolverá _None_ con un mensaje de error.

```python
def Encripta1(Texto,a,b):
    alfabeto=''.join([chr(c) for c in [65..90]])
    simbolos=u'¡'+''.join([chr(c) for c in [33]+[36..62]])+u'¿?'
    alfabeto=' '+alfabeto+alfabeto.lower()+simbolos
    OO,oo=alfabeto.index('O'),alfabeto.index('o')
    alfabeto=alfabeto[:OO]+u'Ñ'+alfabeto[OO:oo]+u'ñ'+alfabeto[oo:]+u'\n'
    L,M=len(alfabeto),len(Texto)
    ### Si el texto tiene longitud impar,
    ### repetimos el último caracter.
    if gcd(L,(a^2-b^2))!=1:
        print('Error, no se puede encriptar con %d y %d.'%(a,b))
        return None    
    if M%2!=0:
        Texto+=Texto[-1]
        M+=1
    encriptado=u''
    for j in xsrange(1,M,2):
        i1,i2=alfabeto.index(Texto[j-1]),alfabeto.index(Texto[j])
        j1,j2=(a*i1+b*i2)%L,(b*i1+a*i2)%L
        encriptado=encriptado+alfabeto[j1]+alfabeto[j2]    
    return encriptado
```

En el c&oacute;digo, las cadenas <span style="font-family: comic sans ms,sans-serif;">vocales</span> y <span style="font-family: comic sans ms,sans-serif;">consonantes</span> se asignar&aacute;n internamente.</p>

```python
def cifradoVC(texto,clave):
    k,t=clave
    vocales='aeiouáéíóú'
    vocales=vocales.upper()+vocales
    consonantes='BCDFGHJKLMNÑPQRSTVWXZ'
    consonantes+=consonantes.lower()
    ###
    L,v,c=len(texto),len(vocales),len(consonantes)
    ####
    tt=u''
    for letra in texto:
        if letra in vocales:
            z=vocales[vocales.index(letra)*k%v]
        elif letra in consonantes:
            z=consonantes[(consonantes.index(letra)+t)%c]
        else:
            z=letra
        tt+=z                        
    return tt
```

<p style="text-align: left;">que tiene un espacio en blanco como primer caracter.</p>

```python
def CifradoTercerParcial(texto,clave,alfabeto=' ABCDEFGHIJKLMNÑOPQRSTUVWXYZabcdefghijklmnñopqrstuvwxyz'):
    L,lc,ltext=len(alfabeto),len(clave),len(texto)
    cifrado=''  
    for i in xsrange(lc):
        cifrado+=alfabeto[(alfabeto.index(texto[i])+alfabeto.index(clave[i]))%L]
    for i in xsrange(lc,ltext):
        cifrado+=alfabeto[(alfabeto.index(texto[i])+alfabeto.index(texto[i-lc]))%L]
    return cifrado
```

En este ejercicio trabajaremos con un sistema de encriptado que no es de sustitución, como el sistema de César, sino de _trasposición_, esto es, basado en reordenar las letras del texto. Se trata de lo siguiente:

La clave está formada por una colección $[n_0, \ldots , n_{k-1}  ]$ de $k$ números enteros. Lo primero que se hace es añadir, si es necesario caracteres adicionales al final del texto (por ejemplo, X) para que la longitud de este "texto ampliado" sea múltiplo de $k$ (por ejemplo, si el texto a codificar tiene 32 caracteres y $k$ vale 5, el texto ampliado contiene, añadido al final del texto original, tres caracteres X). Es con este texto ampliado con el que vamos a proceder a partir de aquí.

Los caracteres que ocupan las posiciones
$$
0, k, 2k, \ldots
$$
del texto ampliado, se reordenan según la regla **desplazar $n_0$ lugares** (módulo la longitud del texto ampliado). Es decir, el carácter en la posición 0 del texto cifrado es el que está en la posición $n_0*k$ del texto ampliado original, el carácter en la posición $k$ del texto cifrado es el que está en la posición $k+n_0*k$ en el texto ampliado original y, para este bloque de posiciones en general la regla es que el carácter que ocupa la posición $j*k$ del texto codificado es el que ocupaba la posición $j*k+n_0*k$ en el texto ampliado original (pensado módulo la longitud del texto ampliado, para que la cosa tenga sentido si  $j*k+n_0*k$ es mayor que dicha longitud).

Del mismo modo, los caracteres que ocupan las posiciones
$$
1, 1+k, 1+2k, \ldots
$$
del texto ampliado original, se reordenan según la regla **desplazar $n_1$ lugares** (módulo la longitud del texto ampliado). Es decir, el carácter que ocupa la posición $1+j*k$ del texto codificado es el que ocupaba la posición $1+j*k+n_1*k$ en el texto ampliado original, módulo la longitud del texto ampliado.

Así hasta el último bloque de posiciones, que son 
$$
(k-1), (k-1)+k, (k-1)+2k, \ldots
$$
Los caracteres en esas posiciones se reordenan según la regla **desplazar $n_{k-1}$ lugares** (módulo la longitud del texto ampliado: es decir, el carácter que ocupa la posición $(k-1)+j*k$ del texto codificado es el que ocupaba la posición $(k-1)+j*k+n_{k-1}*k$ en el texto ampliado original, módulo la longitud del texto ampliado.

**a)** Crea una función $\texttt{codifica}(\texttt{texto,clave})$ que dependa de un texto (cadena de caracteres) y de una clave (una lista), que devuelva el texto codificado según este sistema. Si tu función está correctamente diseñada, la celda de comprobación que tienes debajo debe dar $\texttt{True}$. 

```python
def codifica(texto, clave):
    l=len(clave)
    while len(texto)%l!=0:
        texto+='X'
    textocifrado=''
    for j in xsrange(len(texto)):
        salto=clave[j%l]
        nuevaletra=texto[(j+salto*l)%len(texto)]
        textocifrado+=nuevaletra
    return textocifrado
```

```python
#CELDA DE COMPROBACIÓN
textoprueba='Cuidaos mucho en navidades: ¡Mascarilla, distancia social y pocos contactos!'
claveprueba=[3,2,5]
codifica(textoprueba, claveprueba)=='u nocin da :dvMeac sia\xa1aasdlrt,lci  ascinlsa ioc a pynooccsstoXtai!ooCXmduhsae'
```
