## **Transformaciones afines** ##


La composición de dos biyecciones es una nueva biyección. En los ejercicios de la sesión hemos presentado dos familias de biyecciones de $\mathbb{Z}_m$: las traslaciones y las homotecias. Son dos familias diferentes, con la aplicación identidad en común (es la traslación por 0, pero también la homotecia por 1), y en ambas familias ocurre que la inversa  de una dada es una biyección de la misma familia (la inversa de una traslación es una traslación, la inversa de una homotecia es una homotecia). Las llamadas *transformaciones afines* son las que se construyen mezclando estas dos familias a base de componer dos transformaciones, una de cada familia. 

Por ejemplo, en $\mathbb{Z}_{12}$ podemos hacer la operación siguiente:

$$
x \to 5 \cdot x \to (5 \cdot x) + 2 \pmod{12} \quad 
$$


Es decir, primero aplicamos al elemento $x$ la homotecia $H$ descrita por Hom(5,12), y luego al resultado obtenido le aplicamos la traslación $T$ dada por 
Tras(2,12).

## Ejercicio 3 ##

Si $f:A\to A$ es una biyección de un conjunto finito $A$ en sí mismo, decimos que cierto $a\in A$ es un _punto fijo_ de $f$ si $f(a)=a$.

En el curso de Matemática Discreta, en segundo, demostrarás que el valor promedio del número de puntos fijos de una biyección es 1, independientemente de cuál sea el número $m$ de elementos de $A$. En este ejercicio vamos a hacer dos experimentos relacionados con esto: un cálculo exacto, y una simulación.

**a)** Define una función $\texttt{fijos}(\texttt{f})$ que dependa de un diccionario que represente una biyección $f$ de un conjunto cualquiera en sí mismo, y que devuelva el número de puntos fijos de $f$.


**b)** Toma $m=8$ y calcula por fuerza bruta el promedio del número de puntos que fijan las biyecciones de $A=\{1,\dots,m\}$ en sí mismo, recorriendo una a una todas las biyecciones.

**c)** Si ahora tomas $m=100$, hay poca esperanza de que Sage pueda reproducir el cálculo exacto del apartado anterior en un tiempo razonable (¿cuántas biyecciones hay ahora?). Sin embargo, podemos hacer lo siguiente:

Estima el promedio de puntos que fijan las biyecciones de $A=\{1,\dots,100\}$ en sí mismo eligiendo aleatoriamente biyecciones y contando cuántos puntos fijan. **Nota**.- Como Permutations(A) es en este caso muy grande, para elegir una aleatoriamente no funciona $\texttt{choice}$, pero $.\texttt{random}\_\texttt{element}()$ sí lo hace.



**d)** Un _desbarajuste_ de $A$ es una aplicación biyectiva de $A$ en sí mismo que no fija ningún punto. Estima experimentalmente cuál es la probabilidad de que una biyección de $A=\{1,\dots,m\}$ en sí mismo sea un desbarajuste para $m=10$, $20$, ..., $100$. Después, calcula el valor de $1/e$ (aquí $e$ denota, como es habitual, la base del logaritmo natural).
