## **Transformaciones afines** ##


La composición de dos biyecciones es una nueva biyección. En los ejercicios de la sesión hemos presentado dos familias de biyecciones de $\mathbb{Z}_m$: las traslaciones y las homotecias. Son dos familias diferentes, con la aplicación identidad en común (es la traslación por 0, pero también la homotecia por 1), y en ambas familias ocurre que la inversa  de una dada es una biyección de la misma familia (la inversa de una traslación es una traslación, la inversa de una homotecia es una homotecia). Las llamadas *transformaciones afines* son las que se construyen mezclando estas dos familias a base de componer dos transformaciones, una de cada familia. 

Por ejemplo, en $\mathbb{Z}_{12}$ podemos hacer la operación siguiente:

$$
x \to 5 \cdot x \to (5 \cdot x) + 2 \pmod{12} \quad 
$$


Es decir, primero aplicamos al elemento $x$ la homotecia $H$ descrita por Hom(5,12), y luego al resultado obtenido le aplicamos la traslación $T$ dada por 
Tras(2,12).

```python
def Tras(k,m):
    return dict([(x,(x+k)%m) for x in xsrange(m)])

def Hom(k,m):
    return dict([(j,(j*k)%m) for j in xsrange(m)])

af=dict([(k, Tras(2,12)[Hom(5,12)[k]]) for k in xsrange(12)])
af
```

```python
#Creamos la lista de biyecciones afines
m=8
Lista_Biyecciones_Afines=[]
for j in xsrange(m): #en realidad basta tomar los j's coprimos con m, es decir tales que mcm(j,m)=1 ---en sage el mcm se escribe gcd
    for k in xsrange(m):
        diccionario=dict([(x,Tras(k,m)[Hom(j,m)[x]]) for x in xsrange(m)])
        imagen=len(set(diccionario.values()))
        if imagen==m and diccionario not in Lista_Biyecciones_Afines:
            Lista_Biyecciones_Afines.append(diccionario)
Lista_Biyecciones_Afines, len(Lista_Biyecciones_Afines)
```

## Ejercicio 3 ##

Si $f_A\to A$ es una biyección de un conjunto finito $A$ en sí mismo, decimos que cierto $a\in A$ es un _punto fijo_ de $f$ si $f(a)=a$.

En el curso de Matemática Discreta, en segundo, demostrarás que el valor promedio del número de puntos fijos de una biyección es 1, independientemente de cuál sea el número $m$ de elementos de $A$. En este ejercicio vamos a hacer dos experimentos relacionados con esto: un cálculp exacta, y una simulación.

**a)** Define una función _fijos(f)_ que dependa de un diccionario que represente una biyección f de un conjunto cualquiera en sí mismo, y que devuelva el número de puntos fijos de $f$.


```python
def fijos(f):
    return sum([f[k]==k for k in f])
```

**b)** Toma $m=8$, $A=[1,...,m]$ y calcula por fuerza bruta el promedio del número de puntos que fijan las biyecciones de $A$ en sí mismo, recorriendo una a una todas las biyecciones.

```python
m=8
A=srange(1,m+1)
L=Permutations(A)
total_fijos=0
for permu in L:
    f=dict(zip(A, permu))
    total_fijos+=fijos(f)
N(total_fijos/factorial(m))
```

**c)** Si ahora tomas $m=100$, hay poca esperanza de que Sage pueda reproducir el cálculo exacto del apartado anterior en un tiempo razonable (¿cuántas biyecciones hay ahora?). Sin embargo, podemos hacer lo siguiente:

Estima el promedio de puntos que fijan las biyecciones de A=[1,...,100] en sí mismo eligiendo aleatoriamente biyecciones y contando cuántos puntos fijan. **Nota**.- Como Permutations(A) es en este caso muy grande, para elegir una aleatoriamente no funciona *choice*, pero _random_element()_ sí lo hace.



```python
m=100
A=srange(1,m+1)
L=Permutations(A)
estima_fijos=0
reps=10^5
for _ in xsrange(reps):
    f=dict(zip(A,L.random_element()))
    estima_fijos+=fijos(f)
N(estima_fijos/reps)
```

**d)** Un _desbarajuste_ de $A$ es una aplicación biyectiva de $A$ en sí mismo que no fija ningún punto. Estima experimentalmente cuál es la probabilidad de que una biyección de $A=[1,...,m]$ en sí mismo sea un desbarajuste para m=10,20,...,100. Después, calcula el valor de $1/e$ (aquí $e$ denota, como es habitual, la base del logaritmo natural).

(También serás capaz de demostrar en Matemática Discreta esto que estás viendo aquí).

```python
j=10
for k in xsrange(1,11):
    A=srange(1,k*j+1)
    L=Permutations(A)
    exitos=0
    reps=10^5*k #subo el número de repeticiones para compensar que L es cada vez más grande 
    #(aunque el factorial crece más rápido que una potencia, pero bueno...)
    for _ in xsrange(reps):
        f=dict(zip(A,L.random_element()))
        if fijos(f)==0:
            exitos+=1
    print(f'Para m={k*j} la probabilidad estimada de ser un desbarajuste es {N(exitos/reps)}')
```
