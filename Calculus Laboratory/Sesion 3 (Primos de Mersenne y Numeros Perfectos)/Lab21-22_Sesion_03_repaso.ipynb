**3. Problema 357 de Proyecto Euler**

Considérense los divisores de 30: 1,2,3,5,6,10,15,30. Es fácil comprobar que para cualquier
divisor  $\texttt{d}$ de 30 el entero $\texttt{d+30/d}$ es primo.
Encontrar la suma de todos los enteros positivos menores que 10 millones y tales que para todo divisor $\texttt{d}$
de $\texttt{n}$ se tiene que $\texttt{d+n/d}$ es primo.

*Sugerencias:* (i) Dado que hay que comprobar una propiedad muchas veces, conviene definir una función que haga esta
tarea.

(ii) Nótese, para ahorrar trabajo, que si $\texttt{d}$ es  divisor de $\texttt{n}$, entonces $\texttt{n/d}$ también lo es.

(iii) Obsérvese, también para disminuir el tiempo de cálculo, que,  puesto que 1 es siempre divisor de $\texttt{n}$, y $\texttt{1+n/1=n+1}$
ha de ser primo, podemos restringirnos a los números *inmediatamente anteriores a un primo*.
